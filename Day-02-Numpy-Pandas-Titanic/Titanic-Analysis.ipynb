Import libraries and load dataset

```python
import numpy as np
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
df.head()
```

```python
marks = np.array([70, 80, 90, 60, 100])

print("Marks:", marks)
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
```

```python
print("First five rows:")
display(df.head())

print("Last five rows:")
display(df.tail())

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
display(df.describe())

print("Missing values in each column:")
display(df.isnull().sum())
```

```python
print("Selected passenger details:")
display(df.loc[:, ["Name", "Sex", "Age", "Survived"]].head(10))

print("First five rows and first four columns:")
display(df.iloc[0:5, 0:4])
```

Question 1:

```python
# Question 1:
# How many female passengers in 2nd Class survived?

q1 = df.loc[
    (df["Sex"] == "female") &
    (df["Pclass"] == 2) &
    (df["Survived"] == 1)
]

print("Female passengers in 2nd Class who survived:", len(q1))
display(q1[["Name", "Age", "Pclass", "Survived"]])
```

Question 2:

```python
# Question 2:
# How many passengers younger than 12 survived?
# What were their passenger classes?

q2 = df.loc[
    (df["Age"] < 12) &
    (df["Survived"] == 1)
]

print("Passengers under 12 who survived:", len(q2))

print("\nSurvivors by passenger class:")
display(q2["Pclass"].value_counts().sort_index())
```

Findings

```python
print("FINDINGS")
print("-" * 40)

print(
    f"1. Female passengers in 2nd Class who survived: {len(q1)}"
)

print(
    f"2. Passengers younger than 12 who survived: {len(q2)}"
)

print("\nClass-wise counts for surviving passengers under 12:")
print(q2["Pclass"].value_counts().sort_index().to_string())

print("\nConclusion:")
print(
    "These queries helped me practise filtering rows with multiple "
    "conditions, counting records, selecting columns, and grouping results."
)
```
